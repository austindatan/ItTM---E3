#### Extract Text Data from a PDF file given the assign pages. (p23, 53)

```python
import pdfplumber
import nltk
nltk.download('punkt', quiet=True)
nltk.download('punkt_tab', quiet=True)

from nltk.tokenize import word_tokenize, sent_tokenize

DATASET = "Fin_labExer1_dataset.pdf"
ASSIGNED_PAGES = [23, 53]
extracted_text = ""

with pdfplumber.open(DATASET) as pdf:
    for page_num in ASSIGNED_PAGES:
        page = pdf.pages[page_num - 1]
        text = page.extract_text()
        if text:
            extracted_text += text + "\n\n"

print(f"Successful text extract!")
print(f"Pages: {ASSIGNED_PAGES}")
```

#### Display a sample text data from PDF file.

```python
print("=" * 60)
print("SAMPLE TEXT FROM PDF")
print("=" * 60)
print(extracted_text[:625])
```

#### Perform a word level tokenization.

```python
word_tokens = word_tokenize(extracted_text)

print("=" * 60)
print("WORD LEVEL TOKENIZATION")
print("=" * 60)
print(word_tokens[:5])
```

#### Perform sentence level tokenization.

```python
sentence_tokens = sent_tokenize(extracted_text)

print("=" * 60)
print("SENTENCE  LEVEL TOKENIZATION")
print("=" * 60)
for i, sent in enumerate(sentence_tokens[:3]):
    print(f"[{i + 1}] {sent}")
    print()
```

#### Perform paragraph level tokenization

```python
lines = [line for line in extracted_text.split('\n') if line.strip()]
body_lines = [l for l in lines if not ('A/75/592' in l or '/91' in l)]

paras = []
curr = []
for line in body_lines:
    if curr and (curr[-1].endswith(('.', '!', '?', ',')) and len(curr[-1]) < 65 or line.startswith('Mr. President')):
        paras.append(" ".join(curr))
        curr = []
    curr.append(line)
if curr:
    paras.append(" ".join(curr))

print("=" * 60)
print("PARAGRAPH LEVEL TOKENIZATION")
print("=" * 60)
for i, sent in enumerate(paras[:1]):
    print(f"[{i + 1}] {sent}")
    print()
```

#### Display the number of words, sentences and paragraphs.

```python
num_words = len(word_tokens)
num_sentences = len(sentence_tokens)
num_paragraphs = len(paras)

print("=" * 60)
print("SUMMARY")
print("=" * 60)
print(f"  Pages extracted  : {ASSIGNED_PAGES}")
print(f"  Number of words  : {num_words}")
print(f"  Number of sentences : {num_sentences}")
print(f"  Number of paragraphs: {num_paragraphs}")
```
